## 2 Plotting
Make a scatter plot of **exoplanet Earth radius vs. equilibrium temperatures**. Include units, x/y axis labels, and a legend.

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))
plt.scatter(planet_data['pl_eqt'], planet_data['pl_rade'], alpha=0.7, label='Exoplanets', color='royalblue')


plt.xlabel('Equilibrium Temperature (K)')
plt.ylabel('Planet Radius (Earth Radii)')
plt.title('Exoplanet Radius vs. Equilibrium Temperature')
plt.legend()
plt.grid(True)
plt.tight_layout()


plt.show()
```

## 3 A "Cool" Planet
What planet has the **lowest** equilibrium temperature? Print the planet name and temperature with your `planet_data` variable.

```python

filtered_data = planet_data.dropna(subset=['pl_eqt'])


coldest_planet = filtered_data.loc[filtered_data['pl_eqt'].idxmin()]


print(f"Planet with the lowest equilibrium temperature: {coldest_planet['pl_name']}")
print(f"Equilibrium Temperature: {coldest_planet['pl_eqt']} K")
```

Make a new variable that only contains the information for that planet. \
\
*Hint: Use the `.iloc[[#]]` function. Replace # with the proper index.*

```python

coldest_index = filtered_data['pl_eqt'].idxmin()


coldest_planet_data = planet_data.iloc[[coldest_index]]


coldest_planet_data
```

Print the number of columns in the row. 

```python
print(coldest_planet_data.shape[1])
```

Make a new variable that contains only the columns WITHOUT `Nan` values. *Hint: You can use the function `.dropna(axis=1)`.*

```python
coldest_planet_no_nan = coldest_planet_data.dropna(axis=1)
coldest_planet_no_nan
```
